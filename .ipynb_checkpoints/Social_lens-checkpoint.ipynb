```python
# We begin by loading the demo text file. The file contains multiple profiles, each separated by an empty line.
with open("dataset/Social_data.txt",encoding="utf-8")as f:
    data=f.read()

#Printing the variable shows the raw text exactly as it appears in the file.
print(data)
    
```

```python
# The next step is to break the data into individual profile blocks. Each profile in the file is separated 
# by two newline characters, so splitting on that pattern gives us separate units to work with.

chunks=data.split("\n\n")
chunks=[ch for ch in chunks if len (ch)>3]
# print(chunks)
```

```python
# The list contains one chunk per profile. To examine a single profile and better understand the format.
chunks[1]
```

```python
# The goal is to convert each chunk from raw text into a clean Python dictionary. For that, we define a function 
# that trims extra whitespace, separates the lines, removes labels, and returns the meaningful parts.

def parse_chunk(chunk):
    chunk=chunk.strip()
    lines=chunk.split("\n")
    username=lines[0].replace("username:", "")
    posts=lines[1].replace("Posts: ","")
    followers=lines[2].replace("Followers: ", "")
    following=lines[3].replace("Following:","")
    name=lines[4].replace("Name: ","")
    bio=lines[5].replace("Bio: ","")
    type_of_page=lines[6].replace("Page: ","")

    return {"username": username, "name":name, "posts":posts, "followers":followers, "following":following,"bio":bio,"type_of_page":type_of_page}

# check the chunks step by step
print(parse_chunk(chunks[1]))   

# Running the function on a sample chunk gives a structured dictionary containing the name, 
# posts, followers and bio fields extracted from the original text
    
```

```python
all_chunks=[]
for chunk in chunks:
    parsed_chunk =parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
print(all_chunks)    
```

```python
# text file convert to the json file 

import json
s=json.dumps(all_chunks,indent=4)
print(s)
```
